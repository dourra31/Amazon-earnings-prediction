```python
import numpy as np
from sklearn.linear_model import LogisticRegression


X = np.array([
    [7.5, 2],
    [5, 1],
    [8, 2.5],
    [-1, 0],
    [-5, -2],
    [-10, -2],
    [3, 1],
    [-4, -1],
    [6, 3]
])


y = np.array([1, 0, 1, 0, 0, 0, 1, 0, 1])


model = LogisticRegression()
model.fit(X, y)


new_data = np.array([[6, 2]])  
prediction = model.predict(new_data)


if prediction[0] == 1:
    print("Model predicts: Amazon stock will go UP after earnings.")
else:
    print("Model predicts: Amazon stock will go DOWN after earnings.")
```
